```python
from PIL import Image
from pathlib import Path
import time

input_path = Path(r"\PYTHON_projects\DATA\"pic1".webp")
output_path = Path(r"\PYTHON_projects\DATA\picture.jpg")

print("1. Input exists:", input_path.exists())
print("2. Input size MB:", round(input_path.stat().st_size / 1024 / 1024, 2))

print("3. Opening image...")
img = Image.open(input_path)
print("4. Opened")
print("Format:", img.format)
print("Size:", img.size)
print("Mode:", img.mode)

print("5. Forcing image load...")
start = time.time()
img.load()
print("6. Loaded in", round(time.time() - start, 2), "seconds")

print("7. Converting to RGB...")
img = img.convert("RGB")
print("8. Converted")

print("9. Saving...")
start = time.time()
img.save(output_path, format="JPEG", quality=90)
print("10. Saved in", round(time.time() - start, 2), "seconds")

print("Output exists:", output_path.exists())
print("Output path:", output_path)
```
